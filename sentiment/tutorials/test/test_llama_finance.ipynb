# Test Hugging Face Model: `curiousily/Llama-3-8B-Instruct-Finance-RAG:fastest`

This notebook verifies LangChain integration with the Hugging Face router using the fine-tuned finance model.

```python
import os
import sys
from dotenv import load_dotenv

# Ensure current directory is in python path
notebook_dir = os.getcwd()
if notebook_dir not in sys.path:
    sys.path.insert(0, notebook_dir)

# Load environment variables from .env.local
load_dotenv(".env.local")

hf_api_key = os.getenv("HUGGINGFACE_API_KEY", "").strip('"\' ')
hf_model_name = os.getenv("HUGGINGFACE_MODEL_NAME", "curiousily/Llama-3-8B-Instruct-Finance-RAG").strip('"\' ')
hf_base_url = os.getenv("HUGGINGFACE_BASE_URL", "https://router.huggingface.co/v1").strip('"\' ')

print(f"HF Model Name: {hf_model_name}")
print(f"HF Base URL: {hf_base_url}")
print(f"HF API Key exists: {bool(hf_api_key)}")
```

## Initialize LangChain ChatOpenAI

Since the Hugging Face Router API (`https://router.huggingface.co/v1`) is OpenAI-compatible, we can use LangChain's `ChatOpenAI` wrapper. We import it from the modern `langchain_openai` library.

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint

llm_endpoint = HuggingFaceEndpoint(
    repo_id=hf_model_name,
    task="text-generation",
    huggingfacehub_api_token=hf_api_key,
    temperature=0.1, # Note: Hugging Face endpoints require a value > 0.0
    provider="featherless-ai"
)

llm = ChatHuggingFace(llm=llm_endpoint)

print("LLM initialized successfully!")
```

## Run a Simple Query

We query the model with a simple, direct question about finance/RAG.

```python
query = "Explain what Retrieval-Augmented Generation (RAG) is and how it benefits financial analysis in 2 sentences."
print(f"Querying: {query}\n")

try:
    # Attempting standard .invoke method
    response = llm.invoke(query)
    # Some older LangChain wrappers return a string, newer return a BaseMessage
    response_content = response.content if hasattr(response, 'content') else response
    print("--- Response ---")
    print(response_content)
except Exception as e:
    print(f"Error using .invoke: {e}")
    print("Trying alternative calling method...")
    try:
        response = llm(query)
        print("--- Response (legacy call) ---")
        print(response)
    except Exception as e2:
        print(f"Error with legacy call: {e2}")
```
